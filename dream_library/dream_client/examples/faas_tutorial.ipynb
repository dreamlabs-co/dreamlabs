# Fine-tuning tutorial: use the platform's notebook

This notebook is retired. The fine-tuning tutorial, from installing the dream library to running your fine-tuned model in a control loop, is the Fine-tuning page of the platform site, and its notebook is that page's "Download the notebook" button (`fine_tuning_tutorial_joint_space.ipynb`).

Sign in to the platform site, open Fine-tuning in the sidebar, and run that notebook instead. It is generated from the page, so the two always agree, and it builds the observation a fine-tuned model reads (one `state.<part>` per part of your embodiment config's `state`, one `video.<role>` per role of its `video`, and `prompt`) and reads the reply's `per_key`.
